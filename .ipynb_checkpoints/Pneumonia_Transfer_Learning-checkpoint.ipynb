# 🩻 Pneumonia Detection from Chest X-Rays with Transfer Learning
**Author:** Elavarasu K &nbsp;|&nbsp; **Dataset:** [Chest X-Ray Images (Pneumonia) – Kaggle](https://www.kaggle.com/datasets/paultimothymooney/chest-xray-pneumonia)

## Project Overview
A production-style deep learning pipeline that classifies pediatric chest X-rays as **NORMAL** or **PNEUMONIA**.

| Item | Detail |
|---|---|
| Problem type | Binary image classification (medical screening) |
| Data | ~5,856 pediatric chest X-rays (Guangzhou Women & Children's Medical Center), JPEG, grayscale |
| Imbalance | ≈ 1 Normal : 2.9 Pneumonia |
| Approach | Transfer learning (DenseNet121, ImageNet weights) → 2-stage fine-tuning |
| Primary metric | **Recall / Sensitivity** on Pneumonia (missing a sick child is the costly error) + ROC-AUC |
| Explainability | Grad-CAM heatmaps |
| Deployment | Saved Keras model + Gradio demo app (`app.py`) |

### Real-world issues this notebook deliberately handles
1. **The official `val/` folder has only 16 images** → useless for model selection. We re-split train+val ourselves.
2. **Patient-level leakage** → the same child can have several X-rays. We split **by patient ID**, not by image.
3. **Class imbalance** → class weights + threshold tuning instead of blindly using 0.5.
4. **Black-box concern** → Grad-CAM to check the model looks at the lungs, not at text markers or borders.

```python
CFG = {
    "seed": 42,
    "img_size": 224,
    "batch_size": 32,
    "epochs_baseline": 10,
    "epochs_stage1": 10,
    "epochs_stage2": 15,
    "lr_stage1": 1e-3,
    "lr_stage2": 1e-5,
    "unfreeze_last_n": 60,          # DenseNet layers to unfreeze in Stage 2
    "val_fraction": 0.15,
    "backbone": "DenseNet121",
    "classes": ["NORMAL", "PNEUMONIA"],
    "artifacts_dir": "artifacts",
}

# ⚡ QUICK_RUN = True -> CPU-friendly settings (smaller images, fewer epochs, ~1-2 h on a laptop).
#    Set False on a GPU (Kaggle/Colab) for the full-quality run.
QUICK_RUN = not tf.config.list_physical_devices("GPU")
if QUICK_RUN:
    CFG.update({"img_size": 160, "epochs_baseline": 3, "epochs_stage1": 3,
                "epochs_stage2": 4, "unfreeze_last_n": 30})
print("QUICK_RUN:", QUICK_RUN, "| img_size:", CFG["img_size"])

def seed_everything(seed):
    os.environ["PYTHONHASHSEED"] = str(seed)
    os.environ["TF_DETERMINISTIC_OPS"] = "1"
    random.seed(seed); np.random.seed(seed); tf.random.set_seed(seed)
seed_everything(CFG["seed"])

for gpu in tf.config.list_physical_devices("GPU"):
    tf.config.experimental.set_memory_growth(gpu, True)
if tf.config.list_physical_devices("GPU"):
    keras.mixed_precision.set_global_policy("mixed_float16")   # ~2x faster on T4/P100

ART = Path(CFG["artifacts_dir"]); ART.mkdir(exist_ok=True)
json.dump(CFG, open(ART / "config.json", "w"), indent=2)
```

## Phase 2 — Data Acquisition & Folder Audit
**Goal:** download the dataset, locate it regardless of environment (Kaggle / Colab / local PC), and index every image into a **pandas DataFrame** (path, label, original split, patient ID).

**Key detail — patient ID from filenames:**
- Pneumonia: `person1_bacteria_1.jpeg` → patient `person1`, subtype `bacteria`
- Normal: `IM-0115-0001.jpeg` / `NORMAL2-IM-1427-0001.jpeg` → patient `IM-0115` / `NORMAL2-IM-1427`

Working at DataFrame level (instead of `image_dataset_from_directory`) gives us control over splitting, which is how production pipelines are built.

```python
# Option A (Kaggle notebook): add the dataset via "+ Add Input" — it appears under /kaggle/input
# Option B (Colab / local):
#   pip install kaggle  ->  put kaggle.json in ~/.kaggle/
#   kaggle datasets download -d paultimothymooney/chest-xray-pneumonia -p data --unzip
CANDIDATES = ["/kaggle/input/chest-xray-pneumonia/chest_xray",
              "data/chest_xray", "chest_xray"]
DATA_DIR = next((Path(p) for p in CANDIDATES if Path(p, "train").exists()), None)
assert DATA_DIR is not None, "Dataset not found — download it first (see comment above)."
print("Using:", DATA_DIR)
```

```python
def patient_id(fname, label):
    stem = Path(fname).stem
    if label == "PNEUMONIA":
        return stem.split("_")[0]                      # person1
    m = re.match(r"(.*IM-\d+)", stem)                  # IM-0115 / NORMAL2-IM-1427
    return m.group(1) if m else stem

def subtype(fname, label):
    if label == "NORMAL": return "normal"
    s = Path(fname).stem.lower()
    return "bacteria" if "bacteria" in s else ("virus" if "virus" in s else "unknown")

rows = []
for split in ["train", "val", "test"]:
    for label in CFG["classes"]:
        for f in glob.glob(str(DATA_DIR / split / label / "*.jpeg")):
            rows.append({"path": f, "label": label, "orig_split": split,
                         # test/ restarts the personN numbering -> separate ID namespace (see Phase 4 note)
                         "patient": f"{label}_{'test' if split == 'test' else 'trainval'}_{patient_id(f, label)}",
                         "subtype": subtype(f, label)})
df = pd.DataFrame(rows)
df["y"] = (df.label == "PNEUMONIA").astype(int)
print(df.shape)
df.head()
```

```python
audit = df.pivot_table(index="orig_split", columns="label", values="path", aggfunc="count")
audit["total"] = audit.sum(axis=1)
audit["%pneumonia"] = (audit["PNEUMONIA"] / audit["total"] * 100).round(1)
display(audit)
print("Unique patients:", df.patient.nunique(), "| images per patient (mean):", round(len(df)/df.patient.nunique(), 2))
# 🔎 Finding: val/ has just 16 images -> we will rebuild a proper validation set in Phase 4.
```

```python
# Integrity checks a real pipeline must do: corrupt files & duplicates
import hashlib
def file_ok(p):
    img = cv2.imread(p)
    return img is not None
bad = [p for p in df.path.sample(min(len(df), 6000), random_state=0) if not file_ok(p)]
print("Unreadable images:", len(bad))

df["md5"] = df.path.apply(lambda p: hashlib.md5(open(p, "rb").read()).hexdigest())
dups = df[df.md5.duplicated(keep=False)]
print("Exact duplicate images:", len(dups))
df = df.drop_duplicates("md5").reset_index(drop=True)
```

## Phase 4 — Leak-Free, Patient-Level Data Split
**Goal:** build **train / validation / test** sets with no patient appearing in more than one set.

**Decisions:**
- **Test set = official `test/` folder (624 images)** — kept untouched so results are comparable with published work.
- **Train + official val** are pooled, then split with `StratifiedGroupKFold` (groups = patient) → ~85% train / 15% val, class ratio preserved.
- Verify zero patient overlap with assertions (a real unit test).

> 🔎 **Data finding:** on the first run, the leakage assertion fired — 170 `personN` IDs appear in both `train/` and `test/`. Inspection showed that in **all 170 cases the subtype differs** (e.g. `person1_bacteria_*` in train vs `person1_virus_*` in test), and the numbering ranges restart per folder. Conclusion: the IDs are **reused numbers, not the same children**, so test IDs get their own namespace. The check below still guards against real leakage inside train/val and duplicate images.

```python
test_df = df[df.orig_split == "test"].reset_index(drop=True)
pool = df[df.orig_split != "test"].reset_index(drop=True)

n_splits = round(1 / CFG["val_fraction"])
sgkf = StratifiedGroupKFold(n_splits=n_splits, shuffle=True, random_state=CFG["seed"])
tr_idx, va_idx = next(sgkf.split(pool, pool.y, groups=pool.patient))
train_df, val_df = pool.iloc[tr_idx].reset_index(drop=True), pool.iloc[va_idx].reset_index(drop=True)

# Leakage tests
assert not set(train_df.patient) & set(val_df.patient), "Patient leak train/val!"
assert not set(train_df.patient) & set(test_df.patient), "Patient leak train/test!"
assert not set(train_df.md5) & set(test_df.md5), "Duplicate image leak!"

summary = pd.DataFrame({s: [len(d), d.y.mean().round(3), d.patient.nunique()]
                        for s, d in [("train", train_df), ("val", val_df), ("test", test_df)]},
                       index=["images", "pneumonia_rate", "patients"]).T
display(summary)
for name, d in [("train", train_df), ("val", val_df), ("test", test_df)]:
    d.to_csv(ART / f"{name}_split.csv", index=False)
```

```python
classes = np.array([0, 1])
cw = compute_class_weight("balanced", classes=classes, y=train_df.y)
CLASS_WEIGHT = dict(zip(classes, cw))
print("Class weights:", {CFG['classes'][k]: round(v, 3) for k, v in CLASS_WEIGHT.items()})
# Normal gets a higher weight so each Normal mistake costs more during training
```

## Phase 5 — Input Pipeline (`tf.data`) & Augmentation
**Goal:** a fast, scalable loader: read → decode → grayscale→RGB → resize → (augment, train only) → batch → prefetch.

**Augmentation choices are *medically sensible*:**
- ✅ small rotation (±5–7°), small zoom/shift, brightness/contrast jitter → mimic real acquisition variation
- ❌ **no horizontal flip** — the heart sits on the left; flipping creates anatomically impossible images (dextrocardia)
- ❌ no heavy distortions / colour changes

Normalisation is done **inside the model** (`preprocess_input` layer) so the saved model accepts raw 0–255 images — fewer bugs at deployment.

```python
IMG, BS, AUTO = CFG["img_size"], CFG["batch_size"], tf.data.AUTOTUNE

def load_image(path, label):
    raw = tf.io.read_file(path)
    img = tf.io.decode_jpeg(raw, channels=1)            # force grayscale
    img = tf.image.resize(img, (IMG, IMG), antialias=True)
    img = tf.image.grayscale_to_rgb(img)                 # 3-ch for ImageNet models
    return img, tf.cast(label, tf.float32)

augment = keras.Sequential([
    layers.RandomRotation(0.02),                         # ≈ ±7°
    layers.RandomZoom(0.1),
    layers.RandomTranslation(0.05, 0.05),
    layers.RandomContrast(0.1),
    layers.RandomBrightness(0.1, value_range=(0, 255)),
], name="augmentation")

def make_ds(frame, training=False):
    ds = tf.data.Dataset.from_tensor_slices((frame.path.values, frame.y.values))
    if training:
        ds = ds.shuffle(len(frame), seed=CFG["seed"], reshuffle_each_iteration=True)
    ds = ds.map(load_image, num_parallel_calls=AUTO)
    if not training:
        ds = ds.cache()
    ds = ds.batch(BS)
    if training:
        ds = ds.map(lambda x, y: (augment(x, training=True), y), num_parallel_calls=AUTO)
    return ds.prefetch(AUTO)

train_ds, val_ds, test_ds = make_ds(train_df, True), make_ds(val_df), make_ds(test_df)
```

```python
xb, yb = next(iter(train_ds))
fig, axes = plt.subplots(2, 8, figsize=(18, 5))
for ax, im, lab in zip(axes.ravel(), xb[:16], yb[:16]):
    ax.imshow(tf.cast(im, tf.uint8)); ax.set_title(CFG["classes"][int(lab)], fontsize=9); ax.axis("off")
plt.suptitle("Augmented training batch"); plt.tight_layout(); plt.show()
print("Batch:", xb.shape, xb.dtype, "| pixel range:", float(tf.reduce_min(xb)), "-", float(tf.reduce_max(xb)))
```

## Phase 6 — Baseline CNN from Scratch
**Goal:** set a **reference score** with a small custom CNN. Any transfer-learning model must clearly beat it to justify its extra size/complexity — this is how you defend model choices in an interview or design review.

```python
METRICS = [keras.metrics.BinaryAccuracy(name="acc"),
           keras.metrics.AUC(name="auc"),
           keras.metrics.AUC(curve="PR", name="pr_auc"),
           keras.metrics.Recall(name="recall"),
           keras.metrics.Precision(name="precision")]

def callbacks(name, patience=5):
    return [keras.callbacks.EarlyStopping(monitor="val_auc", mode="max", patience=patience, restore_best_weights=True),
            keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.3, patience=2, min_lr=1e-7, verbose=1),
            keras.callbacks.ModelCheckpoint(str(ART / f"{name}.keras"), monitor="val_auc", mode="max", save_best_only=True),
            keras.callbacks.CSVLogger(str(ART / f"{name}_log.csv"))]

def build_baseline():
    inp = keras.Input((IMG, IMG, 3))
    x = layers.Rescaling(1/255.)(inp)
    for f in [32, 64, 128, 256]:
        x = layers.Conv2D(f, 3, padding="same", activation="relu")(x)
        x = layers.BatchNormalization()(x)
        x = layers.MaxPooling2D()(x)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.4)(x)
    out = layers.Dense(1, activation="sigmoid", dtype="float32")(x)
    return keras.Model(inp, out, name="baseline_cnn")

baseline = build_baseline()
baseline.compile(optimizer=keras.optimizers.Adam(1e-3), loss="binary_crossentropy", metrics=METRICS)
baseline.summary()
t0 = time.time()
hist_base = baseline.fit(train_ds, validation_data=val_ds, epochs=CFG["epochs_baseline"],
                         class_weight=CLASS_WEIGHT, callbacks=callbacks("baseline"))
print(f"Baseline training time: {(time.time()-t0)/60:.1f} min")
```

```python
def plot_history(histories, title):
    h = pd.concat([pd.DataFrame(x.history) for x in histories], ignore_index=True)
    fig, ax = plt.subplots(1, 3, figsize=(17, 4))
    for a, m in zip(ax, ["loss", "auc", "recall"]):
        a.plot(h[m], label="train"); a.plot(h[f"val_{m}"], label="val"); a.set_title(m); a.legend()
        if len(histories) > 1:
            a.axvline(len(histories[0].history["loss"]) - 0.5, ls="--", c="gray")
    plt.suptitle(title); plt.tight_layout(); plt.show()
plot_history([hist_base], "Baseline CNN")
```

## Phase 7 — Transfer Learning, Stage 1 (Frozen Backbone)
**Goal:** reuse a **DenseNet121** pretrained on ImageNet as a frozen feature extractor and train only a new classification head.

**Why DenseNet121?** It's the backbone of Stanford's **CheXNet** (radiologist-level pneumonia detection on ChestX-ray14), compact (~8M params) and strong on fine-grained textures.

**Why two stages?** A randomly initialised head produces large gradients; if the backbone were trainable at that point, those gradients would destroy the pretrained features. So: train head first → then gently fine-tune.

```python
# Preprocessing built from serialisable layers (no Lambda) -> the saved model loads cleanly anywhere
def torch_style_preprocess():      # what densenet.preprocess_input does: /255 then ImageNet mean/std
    return keras.Sequential([layers.Rescaling(1/255.),
                             layers.Normalization(mean=[0.485, 0.456, 0.406],
                                                  variance=[0.229**2, 0.224**2, 0.225**2])], name="preprocess")
def identity_preprocess():         # EfficientNet has rescaling built in
    return keras.Sequential([layers.Identity()], name="preprocess")
def caffe_preprocess():            # ResNet50: RGB->BGR + mean subtraction
    return keras.Sequential([layers.Lambda(lambda x: x[..., ::-1]),
                             layers.Normalization(mean=[103.939, 116.779, 123.68], variance=[1, 1, 1])], name="preprocess")

BACKBONES = {
    "DenseNet121":    (keras.applications.DenseNet121,    torch_style_preprocess),
    "EfficientNetB0": (keras.applications.EfficientNetB0, identity_preprocess),
    "ResNet50":       (keras.applications.ResNet50,       caffe_preprocess),
}

def build_tl_model(backbone_name=CFG["backbone"]):
    Base, preprocess = BACKBONES[backbone_name]
    base = Base(include_top=False, weights="imagenet", input_shape=(IMG, IMG, 3))
    base.trainable = False
    inp = keras.Input((IMG, IMG, 3), name="image")       # raw 0-255 pixels
    x = preprocess()(inp)
    x = base(x, training=False)                          # keep BatchNorm in inference mode
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.4)(x)
    x = layers.Dense(128, activation="relu", kernel_regularizer=keras.regularizers.l2(1e-4))(x)
    x = layers.Dropout(0.3)(x)
    out = layers.Dense(1, activation="sigmoid", dtype="float32", name="pneumonia_prob")(x)
    return keras.Model(inp, out, name=f"{backbone_name}_pneumonia"), base

model, base = build_tl_model()
model.compile(optimizer=keras.optimizers.Adam(CFG["lr_stage1"]), loss="binary_crossentropy", metrics=METRICS)
print(f"Trainable params: {sum(np.prod(w.shape) for w in model.trainable_weights):,} / "
      f"{model.count_params():,}")
t0 = time.time()
hist_s1 = model.fit(train_ds, validation_data=val_ds, epochs=CFG["epochs_stage1"],
                    class_weight=CLASS_WEIGHT, callbacks=callbacks("stage1"))
print(f"Stage 1 time: {(time.time()-t0)/60:.1f} min")
```

## Phase 8 — Transfer Learning, Stage 2 (Fine-Tuning)
**Goal:** unfreeze the **last ~60 layers** of DenseNet121 and train with a **100× smaller learning rate** so high-level features adapt from "ImageNet objects" to "lung opacities".

**Good-practice details:**
- BatchNorm layers stay frozen (their running statistics from ImageNet are kept stable — standard Keras guidance).
- Re-compile after changing `trainable` (otherwise the change is ignored).
- Early stopping on `val_auc` with best-weight restore.

```python
base.trainable = True
for layer in base.layers[:-CFG["unfreeze_last_n"]]:
    layer.trainable = False
for layer in base.layers:
    if isinstance(layer, layers.BatchNormalization):
        layer.trainable = False

model.compile(optimizer=keras.optimizers.Adam(CFG["lr_stage2"]), loss="binary_crossentropy", metrics=METRICS)
print(f"Trainable params now: {sum(np.prod(w.shape) for w in model.trainable_weights):,}")
t0 = time.time()
hist_s2 = model.fit(train_ds, validation_data=val_ds, epochs=CFG["epochs_stage2"],
                    class_weight=CLASS_WEIGHT, callbacks=callbacks("stage2", patience=6))
print(f"Stage 2 time: {(time.time()-t0)/60:.1f} min")
plot_history([hist_s1, hist_s2], "DenseNet121 — Stage 1 (frozen) | Stage 2 (fine-tuned)")
```

## Phase 9 — Evaluation & Threshold Tuning
**Goal:** honest evaluation on the untouched test set, with a **decision threshold chosen on validation data**.

**Clinical reasoning for the threshold:**
- **False negative** (sick child sent home) is far worse than a **false positive** (extra review by a radiologist).
- So we pick the threshold on validation that reaches **≥ 95% sensitivity** while maximising specificity — and we report the trade-off openly.
- Choosing the threshold on *test* would be leakage — it is fixed before we touch test.

```python
def predict_probs(m, ds):
    return m.predict(ds, verbose=0).ravel()

y_val, y_test = val_df.y.values, test_df.y.values
p_val, p_test = predict_probs(model, val_ds), predict_probs(model, test_ds)
p_val_base, p_test_base = predict_probs(baseline, val_ds), predict_probs(baseline, test_ds)

# Threshold selection on VALIDATION
fpr, tpr, thr = roc_curve(y_val, p_val)
TARGET_SENS = 0.95
ok = np.where(tpr >= TARGET_SENS)[0]
best = ok[np.argmin(fpr[ok])]
THRESH_CLINICAL = float(thr[best])
f1s = [(t, f1_score(y_val, p_val >= t)) for t in np.linspace(0.05, 0.95, 91)]
THRESH_F1 = float(max(f1s, key=lambda z: z[1])[0])
print(f"Threshold for ≥{TARGET_SENS:.0%} sensitivity: {THRESH_CLINICAL:.3f} | best-F1 threshold: {THRESH_F1:.2f}")
```

```python
def report(y, p, t):
    yp = (p >= t).astype(int)
    tn, fp, fn, tp = confusion_matrix(y, yp).ravel()
    return {"threshold": round(t, 3), "accuracy": accuracy_score(y, yp),
            "sensitivity (recall)": tp / (tp + fn), "specificity": tn / (tn + fp),
            "precision (PPV)": precision_score(y, yp), "NPV": tn / (tn + fn),
            "F1": f1_score(y, yp), "ROC-AUC": roc_auc_score(y, p), "PR-AUC": average_precision_score(y, p),
            "FN": fn, "FP": fp}

results = pd.DataFrame({
    "Baseline CNN @0.5":            report(y_test, p_test_base, 0.5),
    "DenseNet121 @0.5":             report(y_test, p_test, 0.5),
    "DenseNet121 @best-F1":         report(y_test, p_test, THRESH_F1),
    "DenseNet121 @clinical (≥95% sens)": report(y_test, p_test, THRESH_CLINICAL),
}).T
display(results.style.format(precision=3).background_gradient(subset=["ROC-AUC", "sensitivity (recall)"], cmap="Greens"))
results.to_csv(ART / "test_results.csv")
```

```python
fig, ax = plt.subplots(1, 3, figsize=(18, 5))
cm = confusion_matrix(y_test, (p_test >= THRESH_CLINICAL).astype(int))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=CFG["classes"], yticklabels=CFG["classes"], ax=ax[0])
ax[0].set_xlabel("Predicted"); ax[0].set_ylabel("Actual"); ax[0].set_title(f"Confusion matrix @ {THRESH_CLINICAL:.2f}")

for name, p in [("DenseNet121", p_test), ("Baseline", p_test_base)]:
    f, t, _ = roc_curve(y_test, p); ax[1].plot(f, t, label=f"{name} (AUC={roc_auc_score(y_test, p):.3f})")
    pr, rc, _ = precision_recall_curve(y_test, p); ax[2].plot(rc, pr, label=f"{name} (AP={average_precision_score(y_test, p):.3f})")
ax[1].plot([0, 1], [0, 1], "k--"); ax[1].set_title("ROC (test)"); ax[1].set_xlabel("FPR"); ax[1].set_ylabel("TPR"); ax[1].legend()
ax[2].axhline(y_test.mean(), ls="--", c="k", label="no-skill"); ax[2].set_title("Precision-Recall (test)")
ax[2].set_xlabel("Recall"); ax[2].set_ylabel("Precision"); ax[2].legend()
plt.tight_layout(); plt.show()
print(classification_report(y_test, (p_test >= THRESH_CLINICAL).astype(int), target_names=CFG["classes"], digits=3))
```

```python
# Uncertainty: bootstrap 95% CI for test AUC & sensitivity (reviewers always ask "how stable is that number?")
rng = np.random.default_rng(CFG["seed"])
aucs, sens = [], []
for _ in range(1000):
    i = rng.integers(0, len(y_test), len(y_test))
    if len(np.unique(y_test[i])) < 2: continue
    aucs.append(roc_auc_score(y_test[i], p_test[i]))
    sens.append(recall_score(y_test[i], p_test[i] >= THRESH_CLINICAL))
print(f"Test ROC-AUC: {np.mean(aucs):.3f}  (95% CI {np.percentile(aucs, 2.5):.3f}–{np.percentile(aucs, 97.5):.3f})")
print(f"Sensitivity : {np.mean(sens):.3f}  (95% CI {np.percentile(sens, 2.5):.3f}–{np.percentile(sens, 97.5):.3f})")
```

## Phase 10 — Explainability with Grad-CAM
**Goal:** show **where** the model looks. A trustworthy model highlights the lung fields; a "cheating" model highlights text markers, image borders or medical devices.

**How Grad-CAM works:** take gradients of the class **logit** (pre-sigmoid score — the probability saturates at 0/1 and gives blank maps) w.r.t. the last convolutional feature map → average them per channel as importance weights → weighted sum of feature maps → ReLU → upsample onto the X-ray.

```python
def build_gradcam_model(model, last_conv="conv5_block16_concat"):   # last conv block of DenseNet121
    # the nested sub-model that contains the last conv block (skips the 'preprocess' Sequential)
    base = next(l for l in model.layers if isinstance(l, keras.Model) and
                any(x.name == last_conv for x in l.layers))
    conv_out = base.get_layer(last_conv).output
    feat_model = keras.Model(base.inputs, [conv_out, base.output])
    return base, feat_model

_GC = {}
def gradcam(img_uint8, model, target="pred"):
    # Grad-CAM on the LOGIT (pre-sigmoid). Using the probability fails for confident images:
    # at p≈1.00 or 0.00 the sigmoid gradient underflows to ~0 and the heatmap goes blank.
    # target='pred' explains the predicted class; 'pneumonia' always explains pneumonia evidence.
    if id(model) not in _GC:
        _GC[id(model)] = build_gradcam_model(model)
    base, feat_model = _GC[id(model)]
    pre = model.get_layer("preprocess")
    head_layers = model.layers[model.layers.index(base) + 1:]
    dense_out = head_layers[-1]                              # final Dense(1, sigmoid)
    x = tf.convert_to_tensor(img_uint8[None].astype("float32"))
    with tf.GradientTape() as tape:
        conv, y = feat_model(pre(x), training=False)
        tape.watch(conv)
        for l in head_layers[:-1]: y = l(y, training=False)
        logit = tf.cast(tf.matmul(tf.cast(y, dense_out.kernel.dtype), dense_out.kernel) + dense_out.bias, tf.float32)[:, 0]
        score = logit if target == "pneumonia" else tf.where(logit >= 0, logit, -logit)
    grads = tf.cast(tape.gradient(score, conv), tf.float32)
    conv = tf.cast(conv, tf.float32)
    w = tf.reduce_mean(grads, axis=(0, 1, 2))
    cam = tf.nn.relu(tf.reduce_sum(conv[0] * w, axis=-1)).numpy()
    cam = cv2.resize(cam / (cam.max() + 1e-8), (IMG, IMG))
    return cam, float(tf.sigmoid(logit)[0])

def overlay(img, cam, alpha=0.4):
    heat = cv2.applyColorMap(np.uint8(255 * cam), cv2.COLORMAP_JET)[..., ::-1]
    return np.uint8(img * (1 - alpha) + heat * alpha)

def load_for_view(path):
    im = cv2.imread(path, 0); im = cv2.resize(im, (IMG, IMG))
    return np.stack([im] * 3, -1)
```

```python
test_df["prob"] = p_test
test_df["pred"] = (p_test >= THRESH_CLINICAL).astype(int)
picks = pd.concat([
    test_df[(test_df.y == 1) & (test_df.pred == 1)].nlargest(3, "prob").assign(case="True Positive"),
    test_df[(test_df.y == 0) & (test_df.pred == 0)].nsmallest(3, "prob").assign(case="True Negative"),
])
fig, axes = plt.subplots(2, 6, figsize=(20, 7))
for i, (_, r) in enumerate(picks.iterrows()):
    img = load_for_view(r.path); cam, s = gradcam(img, model)
    axes[0, i].imshow(img); axes[0, i].set_title(f"{r.case}\n{r.label}", fontsize=9); axes[0, i].axis("off")
    axes[1, i].imshow(overlay(img, cam)); axes[1, i].set_title(f"P(pneumonia)={s:.2f}", fontsize=9); axes[1, i].axis("off")
plt.suptitle("Grad-CAM for the predicted class: where the model looks"); plt.tight_layout(); plt.savefig(ART / "gradcam_examples.png", dpi=120); plt.show()
```

## Phase 11 — Error Analysis
**Goal:** understand *which* images fail and *why* — this is what turns a Kaggle notebook into an engineering project.

Questions answered:
1. What do the **false negatives** (missed pneumonia) look like? Are they mostly viral (subtle, diffuse)?
2. What do the **false positives** look like? Low-contrast or poorly positioned films?
3. Does error rate differ between **bacterial vs viral** pneumonia?

```python
test_df["error"] = np.where((test_df.y == 1) & (test_df.pred == 0), "FN",
                    np.where((test_df.y == 0) & (test_df.pred == 1), "FP", "correct"))
display(pd.crosstab(test_df.subtype, test_df.error, normalize="index").round(3))

fig, ax = plt.subplots(1, 2, figsize=(13, 4))
sns.histplot(data=test_df, x="prob", hue="label", bins=40, element="step", ax=ax[0])
ax[0].axvline(THRESH_CLINICAL, c="k", ls="--", label="threshold"); ax[0].set_title("Score distribution by true class"); ax[0].legend()
recall_by_type = test_df[test_df.y == 1].groupby("subtype").apply(lambda d: d.pred.mean())
recall_by_type.plot.bar(ax=ax[1], color=["#D1495B", "#EDAE49"]); ax[1].set_title("Sensitivity by pneumonia sub-type"); ax[1].set_ylim(0, 1)
plt.tight_layout(); plt.show()
```

```python
errs = pd.concat([test_df[test_df.error == "FN"].nsmallest(4, "prob"),
                  test_df[test_df.error == "FP"].nlargest(4, "prob")])
if len(errs):
    fig, axes = plt.subplots(2, len(errs), figsize=(3 * len(errs), 6.5))
    axes = np.array(axes).reshape(2, -1)
    for i, (_, r) in enumerate(errs.iterrows()):
        img = load_for_view(r.path); cam, s = gradcam(img, model, target="pneumonia")
        axes[0, i].imshow(img); axes[0, i].set_title(f"{r.error} | {r.subtype}", fontsize=9); axes[0, i].axis("off")
        axes[1, i].imshow(overlay(img, cam)); axes[1, i].set_title(f"p={s:.2f}", fontsize=9); axes[1, i].axis("off")
    plt.suptitle("Hardest mistakes (top: X-ray, bottom: Grad-CAM of pneumonia evidence)"); plt.tight_layout(); plt.savefig(ART / "gradcam_errors.png", dpi=120); plt.show()
```

**Typical findings to write up (verify with your run):**
- Most false negatives are **viral** cases with subtle, diffuse patterns.
- False positives often have **low contrast, rotation, or overlapping soft tissue**.
- If any Grad-CAM focuses outside the lungs (corner text, shoulder, border) → candidate for **lung-field cropping/segmentation** in v2.

## Phase 12 — Model Export, Inference Function & Deployment
**Goal:** package everything needed to use the model outside this notebook:
- `pneumonia_densenet121.keras` – full model (preprocessing included → takes raw 0–255 RGB 224×224)
- `model_card.json` – threshold, metrics, classes, data version, intended use
- `predict()` – one clean function used by the demo app / an API
- `app.py` (separate file) – Gradio web demo with prediction + Grad-CAM

```python
MODEL_PATH = ART / "pneumonia_densenet121.keras"
model.save(MODEL_PATH)

card = {
    "model_name": "Pneumonia classifier (DenseNet121 transfer learning)",
    "version": "1.0.0",
    "classes": CFG["classes"],
    "input": f"RGB uint8 {IMG}x{IMG} (grayscale X-ray replicated to 3 channels); preprocessing inside model",
    "decision_threshold": round(THRESH_CLINICAL, 4),
    "threshold_policy": f"Chosen on validation set for ≥{TARGET_SENS:.0%} sensitivity",
    "test_metrics": {k: (round(float(v), 4) if isinstance(v, (float, np.floating)) else int(v))
                     for k, v in report(y_test, p_test, THRESH_CLINICAL).items()},
    "training_data": "Kaggle paultimothymooney/chest-xray-pneumonia, patient-level re-split of train+val",
    "intended_use": "Research / portfolio demo. Decision support only — NOT for clinical diagnosis.",
    "limitations": ["Pediatric (1-5 yrs) single-centre data", "Binary only (no bacterial/viral split)",
                    "Not validated on adult or external hospital data"],
}
json.dump(card, open(ART / "model_card.json", "w"), indent=2)
print(json.dumps(card, indent=2))
```

```python
reloaded = keras.models.load_model(MODEL_PATH)
THRESH = json.load(open(ART / "model_card.json"))["decision_threshold"]

def predict(image_path_or_array, mdl=reloaded, threshold=THRESH):
    if isinstance(image_path_or_array, (str, Path)):
        img = cv2.imread(str(image_path_or_array), 0)
    else:
        img = np.asarray(image_path_or_array)
        img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY) if img.ndim == 3 else img
    img = cv2.resize(img, (IMG, IMG), interpolation=cv2.INTER_AREA)
    x = np.stack([img] * 3, -1)[None].astype("float32")
    t0 = time.perf_counter()
    prob = float(mdl.predict(x, verbose=0)[0, 0])
    return {"label": CFG["classes"][int(prob >= threshold)], "pneumonia_probability": round(prob, 4),
            "threshold": threshold, "latency_ms": round((time.perf_counter() - t0) * 1000, 1)}

# Sanity check: reloaded model must reproduce notebook predictions
r = test_df.sample(1, random_state=7).iloc[0]
print("True:", r.label, "|", predict(r.path))
assert np.allclose(predict_probs(reloaded, test_ds)[:50], p_test[:50], atol=1e-3), "Reloaded model mismatch!"
print("✅ Export verified")
```

```python
# Optional: lightweight TFLite export for mobile/edge (e.g., rural clinic tablets)
try:
    conv = tf.lite.TFLiteConverter.from_keras_model(reloaded)
    conv.optimizations = [tf.lite.Optimize.DEFAULT]
    tfl = conv.convert()
    (ART / "pneumonia_densenet121.tflite").write_bytes(tfl)
    print(f"TFLite size: {len(tfl)/1e6:.1f} MB vs Keras: {MODEL_PATH.stat().st_size/1e6:.1f} MB")
except Exception as e:
    print("TFLite export skipped:", e)
```

## Phase 13 — Conclusions, Limitations & Next Steps

### Results — quick CPU run (160×160 images, 3+3+4 epochs, laptop CPU, total ≈ 58 min)
Official test set: 618 images (390 Pneumonia / 228 Normal after removing duplicates), untouched until the end.

| Model | Threshold | Accuracy | ROC-AUC | PR-AUC | Sensitivity | Specificity | F1 | Missed pneumonia (FN) | False alarms (FP) |
|---|---|---|---|---|---|---|---|---|---|
| Baseline CNN | 0.50 | 0.628 | 0.853 | 0.886 | 1.000 | 0.004 | 0.771 | 0 | 230 |
| DenseNet121 (fine-tuned) | 0.50 | **0.877** | **0.955** | **0.971** | 0.928 | **0.792** | **0.904** | 28 | 48 |
| DenseNet121 (fine-tuned) | 0.098 (clinical, chosen on val) | 0.850 | 0.955 | 0.971 | **0.984** | 0.623 | 0.891 | **6** | 87 |

- Bootstrap 95% CI: **ROC-AUC 0.955 (0.940–0.970)**, sensitivity at the clinical threshold **0.984 (0.972–0.995)**.
- Validation AUC: baseline 0.944 → DenseNet Stage 1 0.992 → Stage 2 **0.993**.
- The baseline "predicts pneumonia for everyone" on test (specificity 0.4%) — it overfits the train/val distribution; transfer learning generalises far better.
- The clinical threshold cuts missed pneumonia cases from 28 to **6**, at the cost of 39 more false alarms that a radiologist would review.
- Error analysis: all errors on Normal images are false alarms (37.7% of Normals at the clinical threshold); misses are rare for both bacterial (2.1%) and viral (0.7%) cases.
- Val (0.993 AUC) → test (0.955) drop is expected: the official test set has a different class mix (62.6% vs 74.1% pneumonia) and was graded separately.
- Data findings: 62 exact duplicate images removed; `personN` IDs restart in `test/` (170 shared numbers, all with different subtype) → separate ID namespace.
- Export: Keras model 36.4 MB, TFLite 7.6 MB; reloaded model reproduces predictions.
- Grad-CAM (computed on the logit): true positives light up the **lung fields**, not text markers or borders. True negatives focus on the clear upper lungs/clavicle area. False positives are hazy, low-contrast "normal" films where the model sees diffuse opacity; false negatives show only weak evidence near the lower lung edge — subtle cases a radiologist would also flag as borderline.

*A full GPU run (224×224, more epochs) on Kaggle/Colab typically adds a few points of specificity.*

### Key engineering decisions
1. **Patient-level split** → prevents optimistic scores from the same child in train and val.
2. **Class weights + threshold tuned on validation** → model behaviour aligned with clinical cost of errors.
3. **Two-stage fine-tuning with frozen BatchNorm** → stable adaptation of ImageNet features.
4. **No horizontal flips** → augmentation respects anatomy.
5. **Grad-CAM + error analysis** → checks the model isn't learning shortcuts.
6. **Preprocessing inside the saved model + model card** → fewer train/serve mismatches.

### Limitations
- Single-centre, pediatric-only data → **external validation required** before any real use.
- Labels come from image-level grading, not lab confirmation.
- Binary task; a clinician also needs bacterial vs viral distinction.

### Next steps (v2 roadmap)
- 3-class model (Normal / Bacterial / Viral) and k-fold ensembling.
- Lung segmentation (U-Net) → classify only lung region.
- Test-time augmentation; temperature scaling for calibration.
- External validation on RSNA Pneumonia / NIH ChestX-ray14.
- Serve with FastAPI + Docker; monitor input drift (intensity histograms) in production.
